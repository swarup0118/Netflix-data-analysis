```python
#Run this small block next to clean missing values:
df['Director'].fillna('Unknown', inplace=True)
df['Cast'].fillna('Unknown', inplace=True)
df['Country'].fillna('Unknown', inplace=True)
df['Release_Date'].fillna('Unknown', inplace=True)
df['Rating'].fillna('Unknown', inplace=True)
```

```python
import pandas as pd
import numpy as np

# Convert Release_Date to datetime safely
df['Release_Date'] = pd.to_datetime(df['Release_Date'], errors='coerce')

# Extract year and month
df['Year'] = df['Release_Date'].dt.year
df['Month'] = df['Release_Date'].dt.month

# Split Country (multiple countries separated by commas)
df['Country'] = df['Country'].str.split(',')
df = df.explode('Country')
df['Country'] = df['Country'].str.strip()

# Clean Type column (genre/category if multi-valued)
df['Type'] = df['Type'].str.split(',')
df = df.explode('Type')
df['Type'] = df['Type'].str.strip()

# Extract duration info
def extract_duration(x):
    if pd.isnull(x):
        return np.nan
    if 'Season' in x:
        return np.nan
    elif 'min' in x:
        return int(x.split()[0])
    else:
        return np.nan

def extract_seasons(x):
    if pd.isnull(x):
        return np.nan
    if 'Season' in x:
        return int(x.split()[0])
    else:
        return np.nan

df['Duration_Minutes'] = df['Duration'].apply(extract_duration)
df['Seasons'] = df['Duration'].apply(extract_seasons)
```

```python

import matplotlib.pyplot as plt

# Count of Movies vs TV Shows
category_counts = df['Category'].value_counts()

plt.figure(figsize=(6,4))
category_counts.plot(kind='bar', color=['skyblue','salmon'])
plt.title('Distribution: Movies vs TV Shows')
plt.xlabel('Category')
plt.ylabel('Count')
plt.show()
```

```python
# 1. How have Movies vs TV Shows evolved over the years?
unique_titles = df.drop_duplicates(subset='Show_Id')

yearly_trend = unique_titles.groupby(['Year', 'Category']).size().unstack(fill_value=0)
yearly_trend.plot(kind='line', figsize=(10,6), marker='o')

plt.title('Evolution of Movies vs TV Shows (2008–2021)')
plt.xlabel('Year')
plt.ylabel('Number of Titles Released')
plt.grid(True)
plt.show()
```

```python
# 2. Top 10 Genres and Their Popularity Over Time
top_genres = df['Type'].value_counts().head(10).index
genre_trend = df[df['Type'].isin(top_genres)].groupby(['Year', 'Type']).size().unstack(fill_value=0)

genre_trend.plot(kind='line', figsize=(12,6))
plt.title('Top 10 Genres Popularity Trend (2008–2021)')
plt.xlabel('Year')
plt.ylabel('Number of Titles')
plt.legend(title='Genre')
plt.grid(True)
plt.show()
```

```python
# 3. Top Contributing Countries Over the Years
top_countries = df['Country'].value_counts().head(5).index
country_trend = df[df['Country'].isin(top_countries)].groupby(['Year', 'Country']).size().unstack(fill_value=0)

country_trend.plot(kind='area', figsize=(12,6), alpha=0.7)
plt.title('Top 5 Countries Contribution Over Time')
plt.xlabel('Year')
plt.ylabel('Number of Titles')
plt.legend(title='Country')
plt.show()
```

```python
# 4. Rating Distribution by Category
rating_pivot = unique_titles.pivot_table(index='Rating', columns='Category', aggfunc='size', fill_value=0)
rating_pivot.plot(kind='bar', figsize=(10,5))
plt.title('Ratings Distribution by Category')
plt.xlabel('Rating')
plt.ylabel('Number of Titles')
plt.xticks(rotation=45)
plt.show()
```

```python
# 5. Average Duration of Movies

movie_durations = df[df['Category'] == 'Movie']

plt.figure(figsize=(8,5))
plt.hist(movie_durations['Duration_Minutes'].dropna(), bins=30, color='cornflowerblue', edgecolor='black')
plt.title('Distribution of Movie Duration')
plt.xlabel('Duration (minutes)')
plt.ylabel('Count')
plt.show()
```
